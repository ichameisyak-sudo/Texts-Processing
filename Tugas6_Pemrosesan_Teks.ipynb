```python
from sklearn.feature_extraction.text import TfidfVectorizer

stemmer = PorterStemmer()
stop_words = set(stopwords.words("english"))
corpus = []

for summary in data.summaries.to_list():
    tokens = summary.lower().split()
    tokens = [stemmer.stem(token) for token in tokens if token not in stop_words]
    corpus.append(" ".join(tokens))

print(corpus[0])

vectorizer = TfidfVectorizer(sublinear_tf=True, max_features=5000)
vectorizer.fit_transform(corpus)

print("Total features ", vectorizer.get_feature_names_out().shape)
vectorizer.get_feature_names_out()[:1000]
```

```python
import pandas as pd
from IPython.display import HTML, display
from scipy.spatial.distance import cdist
from sklearn.feature_extraction.text import CountVectorizer

TopK = 10
query = "moon space station launch mission"
vectorizer_bin = CountVectorizer(binary=True, max_features=5000)
documents_binarized_features = vectorizer_bin.fit_transform(corpus).toarray()
query_binarized_features = vectorizer_bin.transform([query]).toarray()

similarities = 1 - cdist(
    documents_binarized_features, query_binarized_features, metric="jaccard"
)
top_indices = similarities[:, 0].argsort()[::-1][:TopK]

titles = []
abstracts = []
similarity_scores = []

for doc_index in top_indices:
    titles.append(data.iloc[doc_index]["title"])
    abstracts.append(data.iloc[doc_index]["summary"].replace("\n", " "))
    similarity_scores.append(similarities[doc_index, 0])

results = pd.DataFrame(
    {
        "Title": titles,
        "Abstract/Summary": abstracts,
        "Similarities": similarity_scores,
    }
)

display(HTML(results.to_html()))
```
