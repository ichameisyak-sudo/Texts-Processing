```python
from gensim import corpora
from gensim.models import LdaModel
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

print("English Stopwords Sample:", stopwords.words('english')[:10])
```

```python
import pandas as pd
import requests

url = 'https://api.spaceflightnewsapi.net/v4/articles/'
response = requests.get(url)
data = response.json()
df = pd.DataFrame(data['results'])
df_pre = df['summary'].dropna().reset_index(drop=True)
print(f"Total data artikel: {len(df_pre)}")
```

```python
from nltk.tokenize import word_tokenize
from tqdm import tqdm

stop_words = set(stopwords.words('english'))
sentences = []
for text in tqdm(df_pre):
    tokens = word_tokenize(text.lower())
    filtered_tokens = [w for w in tokens if w.isalnum() and w not in stop_words]
    sentences.append(filtered_tokens)
```

```python
from gensim import corpora

dictionary = corpora.Dictionary(sentences)
corpus = [dictionary.doc2bow(doc) for doc in sentences]

print(f"Jumlah kata unik di dictionary: {len(dictionary)}")
```

```python
from gensim.models import LdaModel

num_topics = 3
lda_model = LdaModel(
    corpus=corpus,
    id2word=dictionary,
    num_topics=num_topics,
    iterations=5000,
    passes=10,
    alpha='auto',
    eta='auto',
    random_state=42
)

for idx, topic in lda_model.print_topics(-1):
    print(f"Topic {idx}: {topic}")
```

```python
for i, row in enumerate(lda_model[corpus]):
    print(f"Article {i}: {row}")
```

```python
from gensim.models import CoherenceModel

coherence_model = CoherenceModel(model=lda_model, texts=sentences, dictionary=dictionary, coherence='c_v')
coherence_score = coherence_model.get_coherence()
print(f"Coherence Score: {coherence_score:.4f}")
```

```python
import pyLDAvis
import pyLDAvis.gensim_models as gensimvis

pyLDAvis.enable_notebook()
vis = gensimvis.prepare(lda_model, corpus, dictionary)
pyLDAvis.display(vis)
```
